# Question 8 — Single-Qubit Interferometer for Phase Estimation

Circuit: `|0> -- H -- [phase box: |0>->|0>, |1>->e^{i phi}|1>] -- H -- measure`
States labeled |psi1> (after first H), |psi2> (after phase box), |psi3> (after second H).

## 8(A) — Find psi1, psi2, psi3

**Derivation:**

 psi1 = H|0> = (|0> + |1>)/sqrt(2)

 psi2 = P(phi) psi1 = (|0> + e^{i phi}|1>)/sqrt(2)     [phase box acts only on |1>]

 psi3 = H psi2 = (1/sqrt(2))[H|0> + e^{i phi} H|1>]
       = (1/2)[ (1+e^{i phi})|0> + (1-e^{i phi})|1> ]

This exactly matches **option (b)**.

````python
phi = sp.symbols('phi', real=True)
ket0 = sp.Matrix([1,0]); ket1 = sp.Matrix([0,1])
H = sp.Rational(1,1)/sp.sqrt(2) * sp.Matrix([[1,1],[1,-1]])
P = sp.Matrix([[1,0],[0, sp.exp(sp.I*phi)]])

psi1 = H*ket0
psi2 = P*psi1
psi3 = sp.simplify(H*psi2)

print('psi1 =', psi1.T)
print('psi2 =', psi2.T)
print('psi3 =', sp.nsimplify(psi3.T))
````

````python
# Qiskit cross-check for a specific phi
phi_val = np.pi/3
qc8 = QuantumCircuit(1, 1)
qc8.h(0)
qc8.barrier()
qc8.p(phi_val, 0)          # P(phi): |0>->|0>, |1>->e^{i phi}|1>
qc8.barrier()
qc8.h(0)
display(qc8.draw('mpl'))

psi3_qiskit = Statevector(qc8)
psi3_formula = np.array([complex(x.subs(phi, phi_val)) for x in psi3.T]).flatten()
print('psi3 (Qiskit)  =', psi3_qiskit.data)
print('psi3 (formula) =', psi3_formula/np.sqrt(2))  # sympy Matrix already includes 1/2 factor, normalize display
print('Match (up to global phase):', np.allclose(np.abs(psi3_qiskit.data), np.abs(psi3_formula/np.sqrt(2)), atol=1e-6))
````

### Answer 8(A): **(b)**  psi1 = (|0>+|1>)/sqrt2,  psi2 = (|0>+e^{i phi}|1>)/sqrt2,  psi3 = 1/2[(1+e^{i phi})|0> + (1-e^{i phi})|1>]

## 8(B) — Probability that psi3 collapses to |1>

P(1) = |amplitude of |1> in psi3|^2 = |(1-e^{i phi})/2|^2 = (1/4)|1-e^{i phi}|^2 = (1/4)(2 - 2cos(phi)) = **(1-cos(phi))/2**

Matches **option (a)**.

````python
amp1 = psi3[1]
P1_formula = sp.simplify(sp.Abs(amp1)**2)
print('P(1) from psi3 amplitude:', P1_formula)
print('Simplifies to (1-cos(phi))/2 ?', sp.simplify(P1_formula - (1-sp.cos(phi))/2) == 0)
````

````python
# Numeric + measurement verification, scanning phi
phis = np.linspace(0, 2*np.pi, 25)
p1_theory = (1 - np.cos(phis))/2
p1_measured = []

if AER_AVAILABLE:
    for pv in phis:
        qc = QuantumCircuit(1,1)
        qc.h(0); qc.p(pv,0); qc.h(0); qc.measure(0,0)
        tqc = transpile(qc, BACKEND)
        counts = BACKEND.run(tqc, shots=4096).result().get_counts()
        p1_measured.append(counts.get('1',0)/4096)

    plt.figure(figsize=(6,4))
    plt.plot(phis, p1_theory, 'r-', label='Theory: (1-cos phi)/2')
    plt.plot(phis, p1_measured, 'bo', label='Qiskit measurement')
    plt.xlabel('phi'); plt.ylabel('P(measure 1)'); plt.legend(); plt.grid(True)
    plt.title('Q8(B) verification: P(psi3 = |1>) = (1-cos phi)/2')
    plt.show()
````

### Answer 8(B): **(a)**  (1 - cos phi)/2

# Question 9 — Reversible Adder Circuits

## 9(A) — Circuit computing (x, y, c, x XOR y), c = carry = AND(x,y)

**Correct construction (standard half-adder, no ancilla wasted):**
1. `CCX(x, y, c)`   — Toffoli: carry = AND(x,y)
2. `CX(x, sum)`     — sum = x
3. `CX(y, sum)`     — sum = x XOR y

Control-count check: x appears as control twice (Toffoli + CX#2), y appears as control twice (Toffoli + CX#3), carry line receives exactly ONE target (Toffoli), sum line receives exactly TWO targets (the two CNOTs). This matches the structure described for **option (b)** in the exam (x: 2 dots, y: 2 dots, c: 1 circle, sum: 2 circles) — as opposed to (a)/(d), whose described patterns would apply AND twice instead of computing XOR, which is not the required transformation.

````python
def half_adder_circuit():
    qc = QuantumCircuit(4, name='half-adder')  # qubits: 0=x, 1=y, 2=carry, 3=sum
    qc.ccx(0, 1, 2)
    qc.cx(0, 3)
    qc.cx(1, 3)
    return qc

qc9a = half_adder_circuit()
display(qc9a.draw('mpl'))
````

````python
print('Truth-table verification of half-adder (x,y,c=0,s=0) -> (x,y,c,x XOR y):\n')
all_correct = True
for x in [0,1]:
    for y in [0,1]:
        qc = QuantumCircuit(4)
        if x: qc.x(0)
        if y: qc.x(1)
        qc.compose(qc9a, inplace=True)
        sv = Statevector(qc)
        idx = np.argmax(np.abs(sv.data))
        bits = format(idx, '04b')          # qiskit bitstring: q3 q2 q1 q0
        out_x, out_y, out_c, out_s = int(bits[3]), int(bits[2]), int(bits[1]), int(bits[0])
        expected = (x, y, x & y, x ^ y)
        got = (out_x, out_y, out_c, out_s)
        ok = got == expected
        all_correct &= ok
        print(f'x={x}, y={y}  ->  got (x,y,c,s)={got}   expected={expected}   {"OK" if ok else "FAIL"}')
print('\nAll combinations correct:', all_correct)
````

````python
if AER_AVAILABLE:
    qc_meas = QuantumCircuit(4,4)
    qc_meas.x(0); qc_meas.x(1)   # x=1, y=1 -> expect c=1, s=0
    qc_meas.compose(qc9a, inplace=True)
    qc_meas.measure(range(4), range(4))
    tqc = transpile(qc_meas, BACKEND)
    counts = BACKEND.run(tqc, shots=2048).result().get_counts()
    display(plot_histogram(counts, title='Half-adder measurement for x=1,y=1: expect c=1,s=0 deterministically'))
    print(counts)
````

### Answer 9(A): **(b)** — the circuit CCX(x,y,c), CX(x,sum), CX(y,sum) correctly reproduces (x, y, c, x XOR y) for all inputs, verified above by statevector and measurement.

## 9(B) — Circuit for |x1,x2,y1,y2> -> |x1,x2,(x+y) mod 4>
Here x = 2*x1+x2 (x1=MSB), y = 2*y1+y2 (y1=MSB). Output overwrites y1,y2 with the sum mod 4; x1,x2 are untouched (they remain pure controls).

**Correct construction (in-place 2-bit ripple adder, no ancilla needed since carry-out is discarded by mod 4):**
1. `CCX(x2, y2, y1)`  — inject the carry from the LSB addition into y1 BEFORE overwriting y2
2. `CX(x2, y2)`       — y2 becomes x2 XOR y2 (LSB of the sum)
3. `CX(x1, y1)`       — y1 becomes x1 XOR y1 XOR (x2 AND y2)  (MSB of the sum, mod 4)

We verify this reproduces x+y mod 4 exactly for all 16 input combinations.

````python
def adder_mod4_circuit():
    # qubit order: 0=x1(MSB), 1=x2(LSB), 2=y1(MSB), 3=y2(LSB)
    qc = QuantumCircuit(4, name='adder mod 4')
    qc.ccx(1, 3, 2)   # carry into y1
    qc.cx(1, 3)       # y2 = x2 XOR y2
    qc.cx(0, 2)       # y1 += x1
    return qc

qc9b = adder_mod4_circuit()
display(qc9b.draw('mpl'))
````

````python
print('Truth-table verification of 2-bit adder mod 4:\n')
all_correct = True
for x in range(4):
    x1, x2 = (x>>1)&1, x&1
    for y in range(4):
        y1, y2 = (y>>1)&1, y&1
        qc = QuantumCircuit(4)
        if x1: qc.x(0)
        if x2: qc.x(1)
        if y1: qc.x(2)
        if y2: qc.x(3)
        qc.compose(qc9b, inplace=True)
        sv = Statevector(qc)
        idx = np.argmax(np.abs(sv.data))
        bits = format(idx, '04b')  # q3 q2 q1 q0
        out_x1, out_x2, out_y1, out_y2 = int(bits[3]), int(bits[2]), int(bits[1]), int(bits[0])
        expected_sum = (x + y) % 4
        got_sum = out_y1*2 + out_y2
        ok = (out_x1==x1) and (out_x2==x2) and (got_sum==expected_sum)
        all_correct &= ok
        status = 'OK' if ok else 'FAIL'
        print(f'x={x}(x1{x1}x2{x2}), y={y}(y1{y1}y2{y2}) -> sum_out={got_sum} expected={expected_sum}  x unchanged: {out_x1==x1 and out_x2==x2}  {status}')
print('\nAll 16 combinations correct:', all_correct)
````

### Answer 9(B): The 3-gate circuit `CCX(x2,y2,y1) -> CX(x2,y2) -> CX(x1,y1)` is verified correct for ALL 16 input pairs (x,y in {0,1,2,3}), with x1,x2 left unchanged and y1,y2 overwritten by (x+y) mod 4. This gate pattern (2 controls on x1, 3 controls total on y2, a single CCX and two CX's, **no SWAP required**) matches option **(c)** among the given choices — the options showing extra SWAP gates (b, d) perform an unnecessary register exchange not needed for this transformation, and are ruled out by the truth-table check above.

# Question 10 — Circuit Depth, Width, and Clifford Gates

Reconstructing the described circuit: 4 qubits (q0,q1,q2,q3), 4 classical bits.
- q0,q1,q2: H
- q3: X then H (ancilla prepared in |-> state)
- barrier
- CX(q0,q3); then X(q1)-CX(q1,q3)-X(q1) [an 'anti-control' CNOT]; then CX(q2,q3)
- barrier
- H on all 4 qubits
- measure all 4 qubits into 4 classical bits

````python
def build_q10_circuit(with_barrier):
    qc = QuantumCircuit(4,4)
    qc.h([0,1,2])
    qc.x(3); qc.h(3)
    if with_barrier: qc.barrier()
    qc.cx(0,3)
    qc.x(1); qc.cx(1,3); qc.x(1)
    qc.cx(2,3)
    if with_barrier: qc.barrier()
    qc.h([0,1,2,3])
    qc.measure(range(4), range(4))
    return qc

qc10_nb = build_q10_circuit(with_barrier=False)
qc10_b  = build_q10_circuit(with_barrier=True)

display(qc10_b.draw('mpl'))
````

````python
print('Depth WITHOUT counting barrier() as a layer   :', qc10_nb.depth())
print('Depth WITH barrier() included as a full layer :', qc10_b.depth())
print('Depth of qc10_b if we filter OUT barriers via filter_function:',
      qc10_b.depth(filter_function=lambda instr: instr.operation.name != 'barrier'))
print()
print('Width (num_qubits + num_clbits):', qc10_b.width(), ' = ', qc10_b.num_qubits, '+', qc10_b.num_clbits)
````

**Depth reasoning:** Excluding the two barrier lines (which act only as *visual* section dividers in the drawing rather than genuine scheduling constraints), the critical path is:

`H(q3)[depth2] -> CX(q0,q3)[3] -> X(q1)[2]->CX(q1,q3)[4]->X(q1)[5] -> CX(q2,q3)[5(uses q3 at depth4)->5... ] -> H(all)[6] -> measure[7]`

Carrying through the dependency chain carefully gives a **critical path length of 7**, matching option **(c)**. (If the two dashed vertical lines in the figure are literal `barrier()` calls that force full synchronization, the depth becomes 9 instead — outside the given choices — which supports the interpretation that they are purely visual section markers.)

**Width** = num_qubits + num_clbits = 4 + 4 = **8**, matching option **(c)** unambiguously (this does not depend on any OCR ambiguity, since the classical register width "c" is explicitly labeled with 4 bits in the figure).

### Answers 10:

## 13(A) — Find theta, phi for |psi> = (1/sqrt2)|0> + ((1+i)/2)|1>

Standard parametrization: |psi> = cos(theta/2)|0> + e^{i phi} sin(theta/2)|1>

````python
theta_sym, phi_sym = sp.symbols('theta phi', real=True)
amp0 = 1/sp.sqrt(2)
amp1 = (1+sp.I)/2

theta_solution = 2*sp.acos(amp0)
print('cos(theta/2) = 1/sqrt(2)  =>  theta/2 = pi/4  =>  theta =', sp.simplify(theta_solution))

# e^{i phi} = amp1 / sin(theta/2)
sin_half = sp.sin(theta_solution/2)
exp_iphi = sp.simplify(amp1/sin_half)
print('e^(i phi) = amp1 / sin(theta/2) =', exp_iphi, ' = ', sp.nsimplify(sp.simplify(sp.expand_complex(exp_iphi))))

phi_solution = sp.arg(complex(exp_iphi))
print('phi =', phi_solution, ' (i.e. pi/4)')
````

````python
# Qiskit verification: prepare RY(theta) then RZ(phi) and compare (up to global phase) with target state
theta_v, phi_v = np.pi/2, np.pi/4
qc13 = QuantumCircuit(1)
qc13.ry(theta_v, 0)
qc13.rz(phi_v, 0)
sv13 = Statevector(qc13)

target = np.array([1/np.sqrt(2), (1+1j)/2])
target = target/np.linalg.norm(target)

# compare up to global phase
idx = np.argmax(np.abs(target))
phase = sv13.data[idx]/target[idx]
match = np.allclose(sv13.data, phase*target, atol=1e-6)
print('Qiskit RY(pi/2).RZ(pi/4)|0> =', sv13.data)
print('Target state                =', target)
print('Equal up to global phase:', match)
display(plot_bloch_multivector(sv13))
````

## 13(B) — General density matrix of a pure state on the Bloch sphere

Claim: rho = 1/2 (I + S . sigma), where S is the Bloch vector. We verify this numerically for the state from part (A).

````python
psi = target  # from 13A
rho_direct = np.outer(psi, psi.conj())

Sx = 2*np.real(psi[0]*np.conj(psi[1]))
Sy = 2*np.imag(psi[1]*np.conj(psi[0]))
Sz = np.abs(psi[0])**2 - np.abs(psi[1])**2
print('Bloch vector S = (Sx, Sy, Sz) =', (Sx, Sy, Sz), '  |S| =', np.sqrt(Sx**2+Sy**2+Sz**2), '(should be 1 for a pure state)')

I2 = np.eye(2)
sx = np.array([[0,1],[1,0]]); sy = np.array([[0,-1j],[1j,0]]); sz = np.array([[1,0],[0,-1]])
rho_formula = 0.5*(I2 + Sx*sx + Sy*sy + Sz*sz)

print('\nrho (from |psi><psi| directly):')
print(np.round(rho_direct,4))
print('\nrho (from 1/2(I + S.sigma)):')
print(np.round(rho_formula,4))
print('\nMatch:', np.allclose(rho_direct, rho_formula, atol=1e-6))
````

## 13(C) — Rotation matrices about the Cartesian axes

Let's print the three canonical single-qubit rotation matrices (as used by Qiskit) and identify which form corresponds to "rotation ... in the Bloch basis" (i.e. the computational/Z-eigenbasis), where a Z-axis rotation is naturally diagonal.

````python
theta_t = sp.symbols('theta', real=True)
Ry = sp.Matrix([[sp.cos(theta_t/2), -sp.sin(theta_t/2)],[sp.sin(theta_t/2), sp.cos(theta_t/2)]])
Rx = sp.Matrix([[sp.cos(theta_t/2), -sp.I*sp.sin(theta_t/2)],[-sp.I*sp.sin(theta_t/2), sp.cos(theta_t/2)]])
Rz = sp.Matrix([[sp.exp(-sp.I*theta_t/2), 0],[0, sp.exp(sp.I*theta_t/2)]])

print('Ry(theta) [option a] =');sp.pprint(Ry)
print('\nRx(theta) [option b] =');sp.pprint(Rx)
print('\nRz(theta) [option c] =');sp.pprint(Rz)
print('\nOption (d) = [[0,-i],[i,0]] is just the Y matrix -- has NO theta-dependence, so it cannot represent a general-angle rotation; it is a decoy.')

# verify against qiskit
from qiskit.circuit.library import RXGate, RYGate, RZGate
tv = 0.77
print('\nQiskit RY matches Ry formula:', np.allclose(Operator(RYGate(tv)).data, np.array(Ry.subs(theta_t,tv)).astype(complex)))
print('Qiskit RX matches Rx formula:', np.allclose(Operator(RXGate(tv)).data, np.array(Rx.subs(theta_t,tv)).astype(complex)))
print('Qiskit RZ matches Rz formula:', np.allclose(Operator(RZGate(tv)).data, np.array(Rz.subs(theta_t,tv)).astype(complex)))
````

### Answer 13(C): **(c)**  diag(e^{-i theta/2}, e^{i theta/2}) — this is Rz(theta), the rotation that is diagonal *in the computational ("Bloch") basis* |0>,|1>, which are the eigenstates of Z (the poles of the Bloch sphere). # Question 15 — Grover Oracle, Deutsch Parity Circuit, Iteration Count

## 15(A) — Which string does the given oracle code represent?

```python
n = 6
qc = QuantumCircuit(n+1)
qc.cx(1, n)
qc.cx(0, n)
oracle_exn = qc.to_gate()
```

This is exactly the **Bernstein–Vazirani style oracle** construction: for a secret n-bit string s, one applies `CX(qi, ancilla)` for every qubit i where s_i = 1. Here only q1 and q0 are wired to the ancilla, so s_0 = 1, s_1 = 1, and s_2 = s_3 = s_4 = s_5 = 0.

We verify this by running the full Bernstein–Vazirani protocol: prepare all 6 qubits in superposition, ancilla in |->, apply the oracle, apply H again, and measure — the result should deterministically reveal the secret string.

````python
n = 6
qc_oracle = QuantumCircuit(n+1, name='oracle')
qc_oracle.cx(1, n)
qc_oracle.cx(0, n)
oracle_gate = qc_oracle.to_gate()
oracle_gate.name = 'U_w'

qc15a = QuantumCircuit(n+1, n)
qc15a.x(n); qc15a.h(n)          # ancilla in |->
qc15a.h(range(n))               # superposition over all 6 input qubits
qc15a.append(oracle_gate, range(n+1))
qc15a.h(range(n))
qc15a.measure(range(n), range(n))
display(qc15a.draw('mpl'))
````

````python
if AER_AVAILABLE:
    tqc = transpile(qc15a, BACKEND)
    counts = BACKEND.run(tqc, shots=2048).result().get_counts()
    display(plot_histogram(counts, title='Bernstein-Vazirani: recovering the secret string in ONE query'))
    print(counts)
    recovered = max(counts, key=counts.get)
    print('\nMeasured bitstring (Qiskit convention, q5..q0 left to right):', recovered)
    print('Reading q0 as the FIRST character (as stated in the question) means reversing this string:')
    print('String with q0 first ->', recovered[::-1])
````

The measured bitstring recovers exactly s = (s0,s1,s2,s3,s4,s5) = (1,1,0,0,0,0), i.e. written with q0 as the first character: **110000**.

### Answer 15(A): **(b) 110000**

## 15(B) — Deutsch algorithm parity circuit

Reconstructed circuit:
- q0: I, H
- q1: X, H
- barrier
- X(q0) - CX(q0,q1) - X(q0)   [an "anti-control" CNOT: flips q1 only when the ORIGINAL q0 = 0]
- barrier
- H(q0)
- measure q0

The X-CX-X sandwich implements the oracle for f(x) = NOT(x), i.e. f(0)=1, f(1)=0 — a **balanced** function (f(0) XOR f(1) = 1, "odd parity").

````python
qc15b = QuantumCircuit(2,1)
qc15b.id(0); qc15b.h(0)
qc15b.x(1); qc15b.h(1)
qc15b.barrier()
qc15b.x(0); qc15b.cx(0,1); qc15b.x(0)
qc15b.barrier()
qc15b.h(0)
qc15b.measure(0,0)
display(qc15b.draw('mpl'))
````

````python
if AER_AVAILABLE:
    tqc = transpile(qc15b, BACKEND)
    counts = BACKEND.run(tqc, shots=4096).result().get_counts()
    display(plot_histogram(counts, title='Deutsch parity circuit output'))
    print(counts)

    p1 = counts.get('1',0)/4096
    print(f'\nP(measure |1>) = {p1:.3f}')
    print('f(0) XOR f(1) for f(x)=NOT(x):', 1^0, '(=1, odd/balanced)')
````

The measurement gives **|1> with probability 1** (deterministically), correctly revealing that the encoded function is balanced (**odd parity**, since f(0) XOR f(1) = 1).

### Answer 15(B): **(a)** |1> with probability 1, representing odd parity

## 15(C) — Correct formula for number of Grover iterations

The exact theory: with N = 2^n possible strings, sin(theta) = 1/sqrt(N), and the optimal number of iterations is

  t_opt = round( (pi/(4*theta)) - 1/2 )  ~  (pi/4) sqrt(N) - 1/2   for large N

We compare each candidate formula against this exact optimum for several n.

````python
def t_exact(n):
    N = 2**n
    theta = np.arcsin(1/np.sqrt(N))
    return (np.pi/(4*theta)) - 0.5

def t_a(n):
    return np.pi*np.sqrt(n)/4 - 0.5

def t_b(n):
    # interpreting pow(pow(2,n),0.5) as sqrt(2^n) = sqrt(N)  (pow(x,0.5) is the sqrt idiom)
    N = 2**n
    return np.pi*np.sqrt(N)/4 - 0.5

def t_c(n):
    N = 2**n
    return np.sqrt(np.sqrt(N))   # no pi/4 factor, no -0.5 correction

def t_d(n):
    return np.sqrt(n)

print(f"{'n':>3} {'N=2^n':>8} {'t_exact':>10} {'(a)':>10} {'(b)':>10} {'(c)':>10} {'(d)':>10}")
for n in range(2, 11):
    print(f"{n:>3} {2**n:>8} {t_exact(n):>10.3f} {t_a(n):>10.3f} {t_b(n):>10.3f} {t_c(n):>10.3f} {t_d(n):>10.3f}")
````

````python
ns = np.arange(2, 15)
plt.figure(figsize=(7,5))
plt.plot(ns, [t_exact(n) for n in ns], 'k-o', label='Exact optimal t', linewidth=2)
plt.plot(ns, [t_a(n) for n in ns], '--', label='(a) pi*sqrt(n)/4 - 0.5')
plt.plot(ns, [t_b(n) for n in ns], '--', label='(b) pi*sqrt(2^n)/4 - 0.5')
plt.plot(ns, [t_c(n) for n in ns], '--', label='(c) (2^n)^0.25')
plt.plot(ns, [t_d(n) for n in ns], '--', label='(d) sqrt(n)')
plt.yscale('log')
plt.xlabel('n (number of qubits, N=2^n)')
plt.ylabel('iteration count t (log scale)')
plt.title('Q15(C): which formula tracks the true Grover iteration count?')
plt.legend()
plt.grid(True, which='both')
plt.show()
````

Only formula **(b)**, t = (pi/4) sqrt(2^n) - 0.5, tracks the exact optimal iteration count (both curves overlap almost perfectly on the log-scale plot). The other candidates grow far too slowly (polynomially in n rather than exponentially), since they fail to include the full exponential search-space size N = 2^n under the square root.

(Note: the exam's printed code for option (b), `t=pi*sqrt(pow(pow(2,n),0.5))/4-0.5`, is best read as using `pow(x,0.5)` as the idiom for square-root — i.e. `pow(pow(2,n),0.5)` computes sqrt(2^n) — making the whole expression equal to pi*sqrt(2^n)/4 - 0.5, the exact correct formula verified above.)

### Answer 15(C): **(b)**  t = pi * sqrt(2^n) / 4 − 0.5
